# Taller Pandas — Café del Eje (cafeterías de especialidad)

```python
import pandas as pd
import numpy as np
```

## 1. Carga y verificación inicial

```python
df_ventas = pd.read_csv("cafeterias_especialidad.csv", index_col=False)
print(df_ventas.shape)
df_ventas.head()
```

## 2. Selección de variables clave

```python
metodo = df_ventas["Metodo_Extraccion"]
print(type(metodo))
metodo.head()
```

```python
df_ventas[["Sede", "Variedad", "Precio_Taza"]].head()
```

## 3. Localización con .loc[]

```python
df_ventas.loc[45]
```

## 4. Depuración de duplicados

```python
print("Duplicados:", df_ventas.duplicated().sum())
df_ventas = df_ventas.drop_duplicates().reset_index(drop=True)
print("Filas tras limpiar:", len(df_ventas))
```

## 5. Calificación ≥ 4.5

```python
df_ventas[df_ventas["Calificacion_Cliente"] >= 4.5]
```

## 6. Filtro con isin

```python
df_ventas[df_ventas["Sede"].isin(["Manizales", "Pereira"])]
```

## 7. Filtro con between

```python
df_ventas[df_ventas["Precio_Taza"].between(7500, 9500)]
```

## 8. Consulta con .query()

```python
df_ventas.query("Sede == 'Bogotá' and Cantidad >= 3")
```

## 9. Filtro compuesto (Geisha & método con "C")

```python
mask = (df_ventas["Variedad"] == "Geisha") & (df_ventas["Metodo_Extraccion"].str.startswith("C"))
df_ventas[mask]
```

## 10. Valores faltantes

```python
print(df_ventas[["Precio_Taza", "Calificacion_Cliente"]].isna().sum())
df_ventas["Precio_Taza"] = df_ventas["Precio_Taza"].fillna(df_ventas["Precio_Taza"].mean())
df_ventas["Calificacion_Cliente"] = df_ventas["Calificacion_Cliente"].fillna(4.0)
print(df_ventas[["Precio_Taza", "Calificacion_Cliente"]].isna().sum())
```

## 11. Columna Total_Venta con apply

```python
def subtotal(fila):
    return fila["Precio_Taza"] * fila["Cantidad"]

df_ventas["Total_Venta"] = df_ventas.apply(subtotal, axis=1)
df_ventas.head()
```

## 12. Renombrar y reordenar

```python
df_ventas = df_ventas.rename(columns={"Metodo_Extraccion": "Metodo", "Tipo_Grano": "Tueste"})
cols = ["Total_Venta"] + [c for c in df_ventas.columns if c != "Total_Venta"]
df_ventas = df_ventas[cols]
df_ventas.head()
```

## 13. Agregaciones por sede

```python
df_ventas.groupby("Sede").agg(
    Calificacion_Promedio=("Calificacion_Cliente", "mean"),
    Total_Venta_Suma=("Total_Venta", "sum"),
)
```

## 14. Función personalizada `rango`

```python
def rango(series):
    return series.max() - series.min()

df_ventas.groupby("Sede")[["Precio_Taza", "Total_Venta"]].agg(rango)
```

## 15. Agrupamiento multinivel

```python
df_ventas.groupby(["Sede", "Tueste"]).agg(
    Calificacion_Promedio=("Calificacion_Cliente", "mean"),
    Total_Venta_Suma=("Total_Venta", "sum"),
)
```
